4. Extraiga datos. Reactivos:

    a. ¿Qué es lo que tiene que extraer de la columna `cantidad_de_empleados`?

    b. ¿Cómo define una expresión regular para extraer la cantidad de trabajadores?

    d. ¿Cuántos trabajadores hay en total? 

```python
# En la parte de acá reemplazamos la columna porque no se pierde info (creo)
df[cdE] = df[cdE].str.lower()
df[cdE] = df[cdE].apply(
    lambda x: re.findall(r'(\d+\.?\d*)\s*trabajadores', x) if isinstance(x, str) else []
)
df[cdE] = df[cdE].apply(lambda x: x[0] if x else None)

# total de trabajadores post extracción
totalTrabajadores = df[cdE].astype(float).sum()
print(f"Total de trabajadores: {totalTrabajadores}")
```

c) determinando el municipio más inseguro
En este inciso realmente sería una tontería analizar la inseguridad con este tipo de datos. La notación booleana para UNA CATEGORÍA de banco inseguro con un 0 y un 1 es demasiado genérica para tomar decisiones o un análisis prudente. De esta manera

```python
# d) calcular el promedio de trabajadores según cats de banco inseguro
df[cdE] = pd.to_numeric(df[cdE], errors='coerce')
meanWorkers = df.groupby(bIn)[cdE].mean()
catLWorkers = meanWorkers.idxmin()
print(f"Cat con menos trabajadores en promedio: Banco inseguro: {catLWorkers}")
```
